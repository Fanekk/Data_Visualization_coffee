```python
import pandas as pd
from pathlib import Path

coffee_survey = df.copy()

# 1. Clean the column names
coffee_survey.columns = (
    coffee_survey.columns
    .str.strip()
    .str.lower()
    .str.replace("'", "", regex=False)           
    .str.replace('’', '', regex=False)           
    .str.replace(r'[^a-z0-9]+', '_', regex=True)
    .str.strip('_')
)

# 2. Grab the coffee tasting columns dynamically
coffee_cols = [col for col in coffee_survey.columns if col.startswith("coffee")]

# 3. Rename dictionary (Removed the missing 'specify' columns)
rename_mapping = {
    "what_is_your_age": "age",
    "how_many_cups_of_coffee_do_you_typically_drink_per_day": "cups",
    "where_do_you_typically_drink_coffee": "where_drink",
    "how_do_you_brew_coffee_at_home": "brew",
    "how_else_do_you_brew_coffee_at_home": "brew_other",
    "on_the_go_where_do_you_typically_purchase_coffee": "purchase",
    "where_else_do_you_purchase_coffee": "purchase_other",
    "what_is_your_favorite_coffee_drink": "favorite",
    "please_specify_what_your_favorite_coffee_drink_is": "favorite_specify",
    "do_you_usually_add_anything_to_your_coffee": "additions",
    "what_else_do_you_add_to_your_coffee": "additions_other",
    "what_kind_of_dairy_do_you_add": "dairy",
    "what_kind_of_sugar_or_sweetener_do_you_add": "sweetener",
    "before_todays_tasting_which_of_the_following_best_described_what_kind_of_coffee_you_like": "style",
    "how_strong_do_you_like_your_coffee": "strength",
    "what_roast_level_of_coffee_do_you_prefer": "roast_level",
    "how_much_caffeine_do_you_like_in_your_coffee": "caffeine",
    "lastly_how_would_you_rate_your_own_coffee_expertise": "expertise",
    "between_coffee_a_coffee_b_and_coffee_c_which_did_you_prefer": "prefer_abc",
    "between_coffee_a_and_coffee_d_which_did_you_prefer": "prefer_ad",
    "lastly_what_was_your_favorite_overall_coffee": "prefer_overall",
    "do_you_work_from_home_or_in_person": "wfh",
    "in_total_much_money_do_you_typically_spend_on_coffee_in_a_month": "total_spend",
    "why_do_you_drink_coffee": "why_drink",
    "other_reason_for_drinking_coffee": "why_drink_other",
    "do_you_like_the_taste_of_coffee": "taste",
    "do_you_know_where_your_coffee_comes_from": "know_source",
    "what_is_the_most_youve_ever_paid_for_a_cup_of_coffee": "most_paid",
    "what_is_the_most_youd_ever_be_willing_to_pay_for_a_cup_of_coffee": "most_willing",
    "do_you_feel_like_youre_getting_good_value_for_your_money_when_you_buy_coffee_at_a_cafe": "value_cafe",
    "approximately_how_much_have_you_spent_on_coffee_equipment_in_the_past_5_years": "spent_equipment",
    "do_you_feel_like_youre_getting_good_value_for_your_money_with_regards_to_your_coffee_equipment": "value_equipment",
    "number_of_children": "number_children"
}

# Apply the new names
coffee_survey = coffee_survey.rename(columns=rename_mapping)

# 4. Filter columns (Removed the missing 'specify' columns)
final_columns = [
    "submission_id", "age", "cups", "where_drink", "brew", "brew_other",
    "purchase", "purchase_other", "favorite", "favorite_specify", 
    "additions", "additions_other", "dairy", "sweetener", "style", 
    "strength", "roast_level", "caffeine", "expertise"
] + coffee_cols + [
    "prefer_abc", "prefer_ad", "prefer_overall", "wfh", "total_spend",
    "why_drink", "why_drink_other", "taste", "know_source", "most_paid",
    "most_willing", "value_cafe", "spent_equipment", "value_equipment",
    "gender", "education_level", "ethnicity_race",
    "employment_status", "number_children", "political_affiliation"
]

# Apply the final filter
coffee_survey = coffee_survey[final_columns]

# 5. Export
filepath = Path("../Datasets/coffee_cleaned.csv")
filepath.parent.mkdir(parents=True, exist_ok=True)
coffee_survey.to_csv(filepath, index=False)

print("Data successfully cleaned and saved!")
```

```python
# Look at the unique answers for each of your 4 variables
print("--- AGE ---")
print(coffee_survey['age'].unique())

print("\n--- SPENDING ---")
print(coffee_survey['total_spend'].unique())

print("\n--- CUPS PER DAY ---")
print(coffee_survey['cups'].unique())

print("\n--- EXPERTISE ---")
print(coffee_survey['expertise'].unique())
```

```python
# 1. Define the exact, logical order we want for our charts
age_order = [
    '<18 years old', 
    '18-24 years old', 
    '25-34 years old', 
    '35-44 years old', 
    '45-54 years old', 
    '55-64 years old', 
    '>65 years old'
]

spend_order = [
    '<$20', 
    '$20-$40', 
    '$40-$60', 
    '$60-$80', 
    '$80-$100', 
    '>$100'
]

cups_order = [
    'Less than 1', 
    '1', 
    '2', 
    '3', 
    '4', 
    'More than 4'
]

# 2. Apply the categorical sorting to the dataframe
coffee_survey['age'] = pd.Categorical(coffee_survey['age'], categories=age_order, ordered=True)
coffee_survey['total_spend'] = pd.Categorical(coffee_survey['total_spend'], categories=spend_order, ordered=True)
coffee_survey['cups'] = pd.Categorical(coffee_survey['cups'], categories=cups_order, ordered=True)

# 3. Ensure expertise is a clean number (it currently has a decimal point like '7.')
coffee_survey['expertise'] = pd.to_numeric(coffee_survey['expertise'], errors='coerce')
```

```python
import plotly.express as px

# --- 1. Age Groups Plot ---
# Count the values, keeping our custom categorical order
age_counts = coffee_survey['age'].value_counts(sort=False).reset_index()
age_counts.columns = ['Age Group', 'Number of Participants']

fig_age = px.bar(
    age_counts, 
    x='Age Group', 
    y='Number of Participants', 
    title="Who took the survey? (Age Distribution)",
    color_discrete_sequence=['#636EFA'] # Feel free to change colors!
)
fig_age.show()


# --- 2. Monthly Spending Plot ---
spend_counts = coffee_survey['total_spend'].value_counts(sort=False).reset_index()
spend_counts.columns = ['Monthly Spend', 'Number of Participants']

fig_spend = px.bar(
    spend_counts, 
    x='Monthly Spend', 
    y='Number of Participants', 
    title="How much do they spend on coffee per month?",
    color_discrete_sequence=['#EF553B']
)
fig_spend.show()


# --- 3. Cups Per Day Plot ---
cups_counts = coffee_survey['cups'].value_counts(sort=False).reset_index()
cups_counts.columns = ['Cups Per Day', 'Number of Participants']

fig_cups = px.bar(
    cups_counts, 
    x='Cups Per Day', 
    y='Number of Participants', 
    title="Daily Coffee Consumption",
    color_discrete_sequence=['#00CC96']
)
fig_cups.show()


# --- 4. Coffee Expertise Plot ---
# Because expertise is a number (1-10), we can use a histogram to show the distribution
fig_expertise = px.histogram(
    coffee_survey, 
    x='expertise', 
    title="Self-Rated Coffee Expertise (1-10)",
    nbins=10,
    color_discrete_sequence=['#AB63FA']
)
# Force the x-axis to show all numbers 1 through 10
fig_expertise.update_layout(xaxis=dict(tickmode='linear', tick0=1, dtick=1))
fig_expertise.show()
```
