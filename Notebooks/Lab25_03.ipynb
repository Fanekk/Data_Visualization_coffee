```python
g=sns.PairGrid(df,hue='species',palette='coolwarm')
g.map_diag(sns.kdeplot)
g.map_offdiag(sns.kdeplot)
```

**Practice exercise 1**

Show, whether:


1.   The petal length is a good predictor of the sepal width regardless of species
2.   The petal length is a good predictor of the sepal width for each species
3. The petal width can be used to recognize the species



```python
sns.lmplot(data=df, x='petal_length', y='sepal_width')
plt.title("Petal Length vs Sepal Width (All Species)")
plt.show()
```

```python
# Separate regression lines for each species
sns.lmplot(data=df, x='petal_length', y='sepal_width', hue='species')
plt.title("Petal Length vs Sepal Width (By Species)")
plt.show()
```

```python
# Boxplot to show distribution and outliers
sns.boxplot(data=df, x='species', y='petal_width')
plt.title("Petal Width by Species")
plt.show()
```

**Practice exercise 2**

Make a plot that will give you a good idea on whether there is a relationship between the petal length and petal width but also an idea about the shape of the distribution of each (one plot).



```python
sns.jointplot(data=df, x='petal_length', y='petal_width', kind = 'reg')
plt.show()
```
